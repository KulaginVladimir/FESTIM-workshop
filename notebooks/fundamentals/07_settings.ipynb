# Параметры расчёта и временной шаг

Основные параметры расчёта задаются объектом `F.Settings`.

Для вводного занятия достаточно рассмотреть:

- абсолютную и относительную точности решателя: `atol`, `rtol`;
- стационарный или нестационарный режим;
- конечное время расчёта;
- фиксированный или адаптивный временной шаг.


## Параметры решателя

`atol` и `rtol` определяют критерии сходимости нелинейного решателя.

Обычно их выбирают с учётом масштаба решаемых величин. Слишком жёсткие значения могут замедлить расчёт, а слишком мягкие — снизить точность.


```python
import festim as F

settings = F.Settings(
    atol=1e-10,
    rtol=1e-8,
)
```

## Стационарный расчёт

Для стационарной задачи временной шаг и конечное время не задаются.


```python
steady_settings = F.Settings(
    atol=1e-10,
    rtol=1e-8,
    transient=False,
)
```

## Нестационарный расчёт с фиксированным шагом

Для нестационарного расчёта задаются:

- `final_time` — конечное время;
- `stepsize` — временной шаг $\Delta t$.

Постоянный шаг удобен для простых задач и проверки результатов.


```python
transient_settings = F.Settings(
    atol=1e-10,
    rtol=1e-8,
    final_time=100.0,
    stepsize=1.0,
)
```

## Адаптивный временной шаг

`F.Stepsize` может автоматически изменять $\Delta t$ в зависимости от сложности решения на текущем шаге:

- `initial_value` — начальный шаг;
- `growth_factor` — увеличение шага при быстрой сходимости;
- `cutback_factor` — уменьшение шага при затруднённой сходимости;
- `target_nb_iterations` — желаемое число итераций решателя;
- `max_stepsize` — максимальный допустимый шаг.


```python
adaptive_settings = F.Settings(
    atol=1e-10,
    rtol=1e-8,
    final_time=100.0,
    stepsize=F.Stepsize(
        initial_value=0.1,
        growth_factor=1.2,
        cutback_factor=0.5,
        target_nb_iterations=4,
        max_stepsize=5.0,
    ),
)
```

## Контроль важных моментов времени

При адаптивном шаге расчёт может перескочить через момент включения источника, изменения температуры или граничного условия.

Параметр `milestones` заставляет расчёт попасть в заданные моменты времени.


```python
stepsize = F.Stepsize(
    initial_value=0.1,
    growth_factor=1.2,
    cutback_factor=0.5,
    target_nb_iterations=4,
    max_stepsize=5.0,
    milestones=[10.0, 20.0, 50.0],
)

settings_with_milestones = F.Settings(
    atol=1e-10,
    rtol=1e-8,
    final_time=100.0,
    stepsize=stepsize,
)
```

## Практическая рекомендация

Временной шаг влияет на точность результата, даже если расчёт успешно сходится.

Проверьте сходимость по времени:

1. выполните расчёт с выбранным шагом;
2. уменьшите шаг;
3. сравните профиль концентрации, поток или интегральную величину.

Если результат почти не изменился, временное разрешение можно считать достаточным.
