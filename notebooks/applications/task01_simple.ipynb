## Полное решение

```python
import festim as F
import numpy as np
import matplotlib.pyplot as plt

model = F.HydrogenTransportProblem()

model.mesh = F.Mesh1D(
    vertices=np.linspace(0.0, 1.0, num=1001),
)

material = F.Material(D_0=1.0, E_D=0.0)

volume = F.VolumeSubdomain1D(
    id=1,
    borders=[0.0, 1.0],
    material=material,
)
left = F.SurfaceSubdomain1D(id=1, x=0.0)
right = F.SurfaceSubdomain1D(id=2, x=1.0)

model.subdomains = [volume, left, right]

H = F.Species("H")
model.species = [H]

model.temperature = 300.0

model.boundary_conditions = [
    F.FixedConcentrationBC(
        subdomain=left,
        value=1.0,
        species=H,
    ),
    F.FixedConcentrationBC(
        subdomain=right,
        value=0.0,
        species=H,
    ),
]

model.settings = F.Settings(
    atol=1e-10,
    rtol=1e-10,
    final_time=0.2,
    stepsize=F.Stepsize(0.05),
)

profile = F.Profile1DExport(
    field=H,
    subdomain=volume,
    times=[0.05, 0.10, 0.20],
)
model.exports = [profile]

model.initialise()
model.run()

for time, concentration in zip(profile.t, profile.data):
    plt.plot(
        profile.x,
        concentration,
        label=f"{time:.2f} с",
    )

plt.xlabel("Координата, м")
plt.ylabel(r"Концентрация H, м$^{-3}$")
plt.legend()
plt.show()
```
